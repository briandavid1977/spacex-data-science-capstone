## Overview of the DataSet

SpaceX has gained worldwide attention for a series of historic milestones. 

It is the only private company ever to return a spacecraft from low-earth orbit, which it first accomplished in December 2010.
SpaceX advertises Falcon 9 rocket launches on its website with a cost of 62 million dollars wheras other providers cost upward of 165 million dollars each, much of the savings is because Space X can reuse the first stage. 


Therefore if we can determine if the first stage will land, we can determine the cost of a launch. 

This information can be used if an alternate company wants to bid against SpaceX for a rocket launch.

This dataset includes a record for each payload carried during a SpaceX mission into outer space.

```python
%pip install pandas

import io
import sys
import sqlite3
import pandas as pd

url = (
    "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/"
    "IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv"
)

if sys.platform == "emscripten":
    from js import fetch
    response = await fetch(url)
    csv_text = await response.text()
    df = pd.read_csv(io.StringIO(csv_text))
else:
    df = pd.read_csv(url)

df = df.dropna(subset=["Date"])

con = sqlite3.connect(":memory:")
df.to_sql("SPACEXTABLE", con, if_exists="replace", index=False)

print("Loaded records:", len(df))
print("Columns:", df.columns.tolist())
```

## Tasks

Now write and execute SQL queries to solve the assignment tasks.

**Note: If the column names are in mixed case enclose it in double quotes
   For Example "Landing_Outcome"**

### Task 1




##### Display the names of the unique launch sites  in the space mission


```python
query = """
SELECT DISTINCT Launch_Site
FROM SPACEXTABLE
ORDER BY Launch_Site;
"""

display(pd.read_sql_query(query, con))
```


### Task 2


#####  Display 5 records where launch sites begin with the string 'CCA' 


```python
query = """
SELECT *
FROM SPACEXTABLE
WHERE Launch_Site LIKE 'CCA%'
ORDER BY Date
LIMIT 5;
"""

display(pd.read_sql_query(query, con))
```

### Task 3




##### Display the total payload mass carried by boosters launched by NASA (CRS)


```python
query = """
SELECT SUM(PAYLOAD_MASS__KG_) AS Total_Payload_KG
FROM SPACEXTABLE
WHERE Customer = 'NASA (CRS)';
"""

display(pd.read_sql_query(query, con))
```

### Task 4




##### Display average payload mass carried by booster version F9 v1.1


```python
query = """
SELECT AVG(PAYLOAD_MASS__KG_) AS Average_Payload_KG
FROM SPACEXTABLE
WHERE Booster_Version = 'F9 v1.1';
"""

display(pd.read_sql_query(query, con))
```

### Task 5

##### List the date when the first succesful landing outcome in ground pad was acheived.


_Hint:Use min function_ 


```python
query = """
SELECT MIN(Date) AS First_Successful_Ground_Landing
FROM SPACEXTABLE
WHERE Landing_Outcome = 'Success (ground pad)';
"""

display(pd.read_sql_query(query, con))
```

### Task 6

##### List the names of the boosters which have success in drone ship and have payload mass greater than 4000 but less than 6000


```python
query = """
SELECT DISTINCT Booster_Version
FROM SPACEXTABLE
WHERE Landing_Outcome = 'Success (drone ship)'
  AND PAYLOAD_MASS__KG_ > 4000
  AND PAYLOAD_MASS__KG_ < 6000
ORDER BY Booster_Version;
"""

display(pd.read_sql_query(query, con))
```

### Task 7




##### List the total number of successful and failure mission outcomes


```python
query = """
SELECT TRIM(Mission_Outcome) AS Mission_Outcome,
       COUNT(*) AS Mission_Count
FROM SPACEXTABLE
GROUP BY TRIM(Mission_Outcome)
ORDER BY Mission_Count DESC;
"""

display(pd.read_sql_query(query, con))
```

### Task 8



##### List all the booster_versions that have carried the maximum payload mass, using a subquery with a suitable aggregate function.


```python
query = """
SELECT DISTINCT Booster_Version, PAYLOAD_MASS__KG_
FROM SPACEXTABLE
WHERE PAYLOAD_MASS__KG_ = (
    SELECT MAX(PAYLOAD_MASS__KG_)
    FROM SPACEXTABLE
)
ORDER BY Booster_Version;
"""

display(pd.read_sql_query(query, con))
```

### Task 9


##### List the records which will display the month names, failure landing_outcomes in drone ship ,booster versions, launch_site for the months in year 2015.

**Note: SQLLite does not support monthnames. So you need to use  substr(Date, 6,2) as month to get the months and substr(Date,0,5)='2015' for year.**


```python
query = """
SELECT substr(Date, 6, 2) AS Month,
       Landing_Outcome,
       Booster_Version,
       Launch_Site
FROM SPACEXTABLE
WHERE substr(Date, 1, 4) = '2015'
  AND TRIM(Landing_Outcome) = 'Failure (drone ship)'
ORDER BY Date;
"""

display(pd.read_sql_query(query, con))
```

### Task 10




##### Rank the count of landing outcomes (such as Failure (drone ship) or Success (ground pad)) between the date 2010-06-04 and 2017-03-20, in descending order.


```python
query = """
SELECT TRIM(Landing_Outcome) AS Landing_Outcome,
       COUNT(*) AS Outcome_Count
FROM SPACEXTABLE
WHERE Date BETWEEN '2010-06-04' AND '2017-03-20'
GROUP BY TRIM(Landing_Outcome)
ORDER BY Outcome_Count DESC;
"""

display(pd.read_sql_query(query, con))
```
